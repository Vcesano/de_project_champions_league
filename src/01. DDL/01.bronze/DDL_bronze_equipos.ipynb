```sql
DROP TABLE IF EXISTS proyecto_champions_league.bronze.equipos_bronze;
CREATE TABLE IF NOT EXISTS proyecto_champions_league.bronze.equipos_bronze (
    temporada STRING,
    equipo_id STRING,
    equipo_nombre STRING,
    sigla STRING,
    pais    STRING,
    codigo_pais STRING,
    fundacion STRING,
    estadio STRING,
    colores STRING,
    _processing_timestamp TIMESTAMP DEFAULT current_timestamp () COMMENT 'Timestamp de procesamiento'
)
USING DELTA
TBLPROPERTIES ('delta.feature.allowColumnDefaults' = 'supported')
COMMENT 'Tabla de equipos - BRONZE'
```

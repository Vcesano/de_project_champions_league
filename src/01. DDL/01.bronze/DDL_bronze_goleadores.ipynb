```sql
DROP TABLE IF EXISTS proyecto_champions_league.bronze.goleadores_bronze;
CREATE TABLE IF NOT EXISTS proyecto_champions_league.bronze.goleadores_bronze (
    temporada STRING,
    jugador_id STRING,
    jugador_nombre STRING,
    jugador_pais STRING,
    jugador_posicion STRING,
    equipo_id STRING,
    equipo_nombre STRING,
    goles STRING,
    asistencias STRING,
    penales STRING,
    _processing_timestamp TIMESTAMP DEFAULT current_timestamp() COMMENT 'Timestamp de procesamiento'
)
USING DELTA
TBLPROPERTIES ('delta.feature.allowColumnDefaults' = 'supported')
COMMENT 'Tabla de goleadores - BRONZE'
```

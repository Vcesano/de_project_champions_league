```sql
CREATE VIEW IF NOT EXISTS proyecto_champions_league.semantica.vw_posiciones AS
SELECT
    temporada,
    grupo,
    posicion,
    eq.nombre nombre_equipo,
    eq.pais pais_equipo,
    jugados,
    ganados,
    empatados,
    perdidos,
    goles_favor,
    goles_contra,
    diferencia, 
    puntos
FROM proyecto_champions_league.gold.fact_posiciones_gold pos
LEFT JOIN proyecto_champions_league.gold.dim_equipos_gold eq
    ON pos.equipo_id = eq.id
ORDER BY temporada, grupo, posicion
```

```sql
-- CREATE OR REPLACE VIEW proyecto_champions_league.semantica.vw_partidos AS
SELECT
    fp.temporada,
    fp.fecha,
    fp.fase,
    fp.grupo,
    fp.jornada,
    eqlocal.nombre nombre_local,
    eqlocal.pais pais_local,
    eqlocal.estadio esadio_local,
    eqvisitante.nombre nombre_visitante,
    eqvisitante.pais pais_visitante,
    arb.arbitro nombre_arbitro,
    arb.arbitro_pais pais_arbitro,
    fp.ganador,
    fp.goles_local,
    fp.goles_visitante,
    fp.goles_local_ht,
    fp.goles_visitante_ht,
    fp.goles_local_et,
    fp.goles_visitante_et,
    fp.goles_local_pen,
    fp.goles_visitante_pen
FROM proyecto_champions_league.gold.fact_partidos_gold fp
LEFT JOIN proyecto_champions_league.gold.dim_equipos_gold eqlocal
    ON fp.local_id = eqlocal.id
LEFT JOIN proyecto_champions_league.gold.dim_equipos_gold eqvisitante
    ON fp.visitante_id = eqvisitante.id
LEFT JOIN proyecto_champions_league.gold.dim_arbitros_gold arb
    ON fp.arbitro_id = arb.id
WHERE
    temporada = 2025
    and (
        eqlocal.nombre = 'Arsenal FC'
        OR
        eqvisitante.nombre = 'Arsenal FC'
    )
```

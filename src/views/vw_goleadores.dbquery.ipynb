```sql
CREATE VIEW IF NOT EXISTS proyecto_champions_league.semantica.vw_goleadores AS
SELECT
    fact.temporada,
    dj.nombre nombre_jugador,
    dj.pais pais_jugador,
    dj.posicion posicion_jugador,
    de.nombre nombre_equipo,
    de.pais pais_equipo,
    fact.goles,
    fact.asistencias,
    fact.penales,
    dense_rank() OVER(PARTITION BY fact.temporada ORDER BY fact.goles DESC, fact.asistencias DESC) ranking
FROM proyecto_champions_league.gold.fact_goleadores_gold fact
LEFT JOIN proyecto_champions_league.gold.dim_jugadores_gold dj
    ON fact.jugador_id = dj.id
LEFT JOIN proyecto_champions_league.gold.dim_equipos_gold de
    ON fact.equipo_id = de.id
```

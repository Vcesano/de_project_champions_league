```python
import requests
import time
import json

API_KEY  = "429daf196cf14483855c7a540ae232bc"
BASE_URL = "https://api.football-data.org/v4"
HEADERS  = {"X-Auth-Token": API_KEY}

TEMPORADAS = [2023, 2024, 2025]   # los únicos con acceso confirmado

print("✅ Config lista")
```

```python
# Endpoint Partidos
filas_partidos = []

for year in TEMPORADAS:
    print(f"\n📅 Partidos Champions League {year}...")

    url      = f"{BASE_URL}/competitions/CL/matches"
    response = requests.get(url, headers=HEADERS, params={"season": year})
    data     = response.json()

    if "matches" not in data:
        print(f"  ⚠️ Error: {data.get('message')}")
        continue

    for m in data["matches"]:
        fila = {
            "match_id":            str(m["id"]),
            "temporada":           str(year),
            "fecha_utc":           str(m["utcDate"]),
            "status":              str(m["status"]),
            "stage":               str(m["stage"]),
            "grupo":               str(m.get("group")),
            "jornada":             str(m.get("matchday")),
            "local_id":            str(m["homeTeam"]["id"]),
            "local_nombre":        str(m["homeTeam"]["name"]),
            "local_sigla":         str(m["homeTeam"].get("tla")),
            "visitante_id":        str(m["awayTeam"]["id"]),
            "visitante_nombre":    str(m["awayTeam"]["name"]),
            "visitante_sigla":     str(m["awayTeam"].get("tla")),
            "goles_local":         str(m["score"]["fullTime"]["home"]),
            "goles_visitante":     str(m["score"]["fullTime"]["away"]),
            "goles_local_ht":      str(m["score"]["halfTime"]["home"]),
            "goles_visitante_ht":  str(m["score"]["halfTime"]["away"]),
            "goles_local_et":      str(m["score"].get("extraTime", {}) and m["score"]["extraTime"].get("home")),
            "goles_visitante_et":  str(m["score"].get("extraTime", {}) and m["score"]["extraTime"].get("away")),
            "goles_local_pen":     str(m["score"].get("penalties", {}) and m["score"]["penalties"].get("home")),
            "goles_visitante_pen": str(m["score"].get("penalties", {}) and m["score"]["penalties"].get("away")),
            "ganador":             str(m["score"].get("winner")),
            "arbitro":             str(m["referees"][0]["name"] if m.get("referees") else None),
            "arbitro_pais":        str(m["referees"][0].get("nationality") if m.get("referees") else None),
        }
        filas_partidos.append(fila)

    print(f"  ✅ {len(data['matches'])} partidos")
    time.sleep(7)

print(f"\nTotal partidos: {len(filas_partidos)}")
```

```python
# Endpoint Goleadores

filas_scorers = []

for year in TEMPORADAS:
    print(f"\n⚽ Goleadores Champions {year}...")

    url      = f"{BASE_URL}/competitions/CL/scorers"
    response = requests.get(url, headers=HEADERS, params={"season": year, "limit": 50})
    data     = response.json()

    if "scorers" not in data:
        print(f"  ⚠️ Error: {data.get('message')}")
        continue

    for s in data["scorers"]:
        fila = {
            "temporada":        str(year),
            "jugador_id":       str(s["player"]["id"]),
            "jugador_nombre":   str(s["player"]["name"]),
            "jugador_pais":     str(s["player"].get("nationality")),
            "jugador_posicion": str(s["player"].get("position")),
            "equipo_id":        str(s["team"]["id"]),
            "equipo_nombre":    str(s["team"]["name"]),
            "goles":            str(s.get("goals")),
            "asistencias":      str(s.get("assists")),
            "penales":          str(s.get("penalties")),
        }
        filas_scorers.append(fila)

    print(f"  ✅ {len(data['scorers'])} goleadores")
    time.sleep(7)

print(f"\nTotal goleadores: {len(filas_scorers)}")
```

```python
# Endpoint Equipos

filas_equipos = []

for year in TEMPORADAS:
    print(f"\n🏴 Equipos Champions {year}...")

    url      = f"{BASE_URL}/competitions/CL/teams"
    response = requests.get(url, headers=HEADERS, params={"season": year})
    data     = response.json()

    if "teams" not in data:
        print(f"  ⚠️ Error: {data.get('message')}")
        continue

    for t in data["teams"]:
        fila = {
            "temporada":     str(year),
            "equipo_id":     str(t["id"]),
            "equipo_nombre": str(t["name"]),
            "sigla":         str(t.get("tla")),
            "pais":          str(t.get("area", {}).get("name")),
            "codigo_pais":   str(t.get("area", {}).get("code")),
            "fundacion":     str(t.get("founded")),
            "estadio":       str(t.get("venue")),
            "colores":       str(t.get("clubColors")),
        }
        filas_equipos.append(fila)

    print(f"  ✅ {len(data['teams'])} equipos")
    time.sleep(7)

print(f"\nTotal equipos: {len(filas_equipos)}")
```

```python
# Endpoint Posiciones

filas_standings = []

for year in TEMPORADAS:
    print(f"\n📊 Standings Champions {year}...")

    url      = f"{BASE_URL}/competitions/CL/standings"
    response = requests.get(url, headers=HEADERS, params={"season": year})
    data     = response.json()

    if "standings" not in data:
        print(f"  ⚠️ Error: {data.get('message')}")
        continue

    for standing in data["standings"]:
        grupo = standing.get("group")
        for row in standing["table"]:
            fila = {
                "temporada":       str(year),
                "grupo":           str(grupo),
                "posicion":        str(row["position"]),
                "equipo_id":       str(row["team"]["id"]),
                "equipo_nombre":   str(row["team"]["name"]),
                "jugados":         str(row["playedGames"]),
                "ganados":         str(row["won"]),
                "empatados":       str(row["draw"]),
                "perdidos":        str(row["lost"]),
                "goles_favor":     str(row["goalsFor"]),
                "goles_contra":    str(row["goalsAgainst"]),
                "diferencia":      str(row["goalDifference"]),
                "puntos":          str(row["points"]),
            }
            filas_standings.append(fila)

    print(f"  ✅ standings cargados")
    time.sleep(7)

print(f"\nTotal filas standings: {len(filas_standings)}")
```

```python
# Creacion de schemas y dataframes

from pyspark.sql.types import StructType, StructField, StringType

schema_partidos = StructType([StructField(c, StringType()) for c in [
    "match_id","temporada","fecha_utc","status","stage","grupo","jornada",
    "local_id","local_nombre","local_sigla","visitante_id","visitante_nombre",
    "visitante_sigla","goles_local","goles_visitante","goles_local_ht",
    "goles_visitante_ht","goles_local_et","goles_visitante_et",
    "goles_local_pen","goles_visitante_pen","ganador","arbitro","arbitro_pais"
]])

schema_scorers = StructType([StructField(c, StringType()) for c in [
    "temporada","jugador_id","jugador_nombre","jugador_pais",
    "jugador_posicion","equipo_id","equipo_nombre","goles","asistencias","penales"
]])

schema_equipos = StructType([StructField(c, StringType()) for c in [
    "temporada","equipo_id","equipo_nombre","sigla","pais",
    "codigo_pais","fundacion","estadio","colores"
]])

schema_standings = StructType([StructField(c, StringType()) for c in [
    "temporada","grupo","posicion","equipo_id","equipo_nombre",
    "jugados","ganados","empatados","perdidos",
    "goles_favor","goles_contra","diferencia","puntos"
]])

df_partidos  = spark.createDataFrame(filas_partidos,  schema_partidos)
df_scorers   = spark.createDataFrame(filas_scorers,   schema_scorers)
df_equipos   = spark.createDataFrame(filas_equipos,   schema_equipos)
df_standings = spark.createDataFrame(filas_standings, schema_standings)

print(f"✅ Partidos:   {df_partidos.count()} filas")
print(f"✅ Goleadores: {df_scorers.count()} filas")
print(f"✅ Equipos:    {df_equipos.count()} filas")
print(f"✅ Standings:  {df_standings.count()} filas")
```

```python
# Validacion de DFs

dfs = [df_partidos, df_scorers, df_equipos, df_standings]

for df in dfs:
  display(df)
```

```python
# Materializacion de DFs en tablas Delta

df_partidos.write.format('delta').mode('overwrite').option('overwriteSchema', 'true').saveAsTable('proyecto_champions_league.bronze.partidos_bronze')
df_scorers.write.format('delta').mode('overwrite').option('overwriteSchema', 'true').saveAsTable('proyecto_champions_league.bronze.goleadores_bronze')
df_equipos.write.format('delta').mode('overwrite').option('overwriteSchema', 'true').saveAsTable('proyecto_champions_league.bronze.equipos_bronze')
df_standings.write.format('delta').mode('overwrite').option('overwriteSchema', 'true').saveAsTable('proyecto_champions_league.bronze.posiciones_bronze')
```

```sql
INSERT INTO proyecto_champions_league.silver.partidos_silver(
    id, temporada, fecha, fase, grupo, jornada, local_id, visitante_id, 
    goles_local, goles_visitante, goles_local_ht, goles_visitante_ht, goles_local_et, goles_visitante_et, goles_local_pen, goles_visitante_pen, 
    ganador, arbitro, arbitro_pais
)
SELECT
    CAST( match_id As INT) AS id,
    CAST( temporada AS INT) AS temporada,
    CAST(fecha_utc AS DATE) AS fecha,
    CASE
        WHEN stage = 'GROUP_STAGE' THEN 'fase de grupos'
        WHEN stage = 'PLAYOFFS' THEN 'playoff'
        WHEN stage = 'LEAGUE_STAGE' THEN 'fase de liga'
        WHEN stage = 'LAST_16' THEN 'octavos de final'
        WHEN stage = 'QUARTER_FINALS' THEN 'cuartos de final'
        WHEN stage = 'SEMI_FINALS' THEN 'semi final'
        WHEN stage = 'FINALS' THEN 'final'
        ELSE 'desconocido'
    END AS fase,
    CASE
        WHEN grupo = 'GROUP_A' THEN 'grupo A'
        WHEN grupo = 'GROUP_B' THEN 'grupo B'
        WHEN grupo = 'GROUP_C' THEN 'grupo C'
        WHEN grupo = 'GROUP_D' THEN 'grupo D'
        WHEN grupo = 'GROUP_E' THEN 'grupo E'
        WHEN grupo = 'GROUP_F' THEN 'grupo F'
        WHEN grupo = 'GROUP_G' THEN 'grupo G'
        WHEN grupo = 'GROUP_H' THEN 'grupo H'
        ELSE 'sin grupo'
    END AS grupo,
    CASE
        WHEN stage = 'FINAL' and jornada IN ('None', '0') THEN 1
        ELSE CAST(jornada as INT)
    END AS jornada,
    CAST(local_id AS INT) AS local_id,
    CAST(visitante_id AS INT) AS visitante_id,
    CAST( goles_local AS INT) AS goles_local,
    CAST( goles_visitante AS INT) AS goles_visitante,
    CASE
        WHEN goles_local_ht = '{}' THEN 0
        ELSE CAST( goles_local_ht AS INT)
    END AS goles_local_ht,
    CASE
        WHEN goles_visitante_ht = '{}' THEN 0
        ELSE CAST( goles_visitante_ht AS INT)
    END AS goles_visitante_ht,
    CASE
        WHEN goles_local_et = '{}' THEN 0
        ELSE CAST( goles_local_et AS INT)
    END AS goles_local_et,
    CASE
        WHEN goles_visitante_et = '{}' THEN 0
        ELSE CAST( goles_visitante_et AS INT)
    END AS goles_visitante_et,
    CASE
        WHEN goles_local_pen = '{}' THEN 0
        ELSE CAST( goles_local_pen AS INT)
    END AS goles_local_pen,
    CASE
        WHEN goles_visitante_pen = '{}' THEN 0
        ELSE CAST( goles_visitante_pen AS INT)
    END AS goles_visitante_pen,
    CASE
        WHEN ganador = 'DRAW' THEN 'empate'
        WHEN ganador = 'HOME_TEAM' THEN 'local'
        WHEN ganador = 'AWAY_TEAM' THEN 'visitante'
        ELSE 'desconocido'
    END AS ganador,
    TRIM(arbitro) AS arbitro,
    CASE
        WHEN arbitro_pais = 'Italy' THEN 'Italia'
        WHEN arbitro_pais = 'France' THEN 'Francia'
        WHEN arbitro_pais = 'Germany' THEN 'Alemania'
        WHEN arbitro_pais = 'Spain' THEN 'España'
        WHEN arbitro_pais = 'England' THEN 'Inglaterra'
        WHEN arbitro_pais = 'Slovenia' THEN 'Eslovenia'
        WHEN arbitro_pais = 'Netherlands' THEN 'Países Bajos'
        WHEN arbitro_pais = 'Romania' THEN 'Rumania'
        WHEN arbitro_pais = 'Poland' THEN 'Polonia'
        WHEN arbitro_pais = 'Portugal' THEN 'Portugal'
        WHEN arbitro_pais = 'Norway' THEN 'Noruega'
        WHEN arbitro_pais = 'Switzerland' THEN 'Suiza'
        WHEN arbitro_pais = 'Sweden' THEN 'Suecia'
        WHEN arbitro_pais = 'Lithuania' THEN 'Lituania'
        WHEN arbitro_pais = 'Bosnia and Herzegovina' THEN 'Bosnia y Herzegovina'
        WHEN arbitro_pais = 'Slovakia' THEN 'Eslovaquia'
        WHEN arbitro_pais = 'Turkey' THEN 'Turquía'
        WHEN arbitro_pais = 'Belgium' THEN 'Bélgica'
        WHEN arbitro_pais = 'Azerbaijan' THEN 'Azerbaiyán'
        WHEN arbitro_pais = 'Scotland' THEN 'Escocia'
        WHEN arbitro_pais = 'Bulgaria' THEN 'Bulgaria'
        WHEN arbitro_pais = 'Ukraine' THEN 'Ucrania'
        WHEN arbitro_pais = 'Georgia' THEN 'Georgia'
        WHEN arbitro_pais = 'Serbia' THEN 'Serbia'
        WHEN arbitro_pais = 'Latvia' THEN 'Letonia'
        WHEN arbitro_pais = 'Montenegro' THEN 'Montenegro'
        WHEN arbitro_pais = 'Croatia' THEN 'Croacia'
        WHEN arbitro_pais = 'Albania' THEN 'Albania'
        WHEN arbitro_pais = 'Denmark' THEN 'Dinamarca'
        WHEN arbitro_pais = 'Austria' THEN 'Austria'
        WHEN arbitro_pais = 'None' THEN 'desconocido'
        ELSE 'desconocido'
    END AS arbitro_pais
FROM proyecto_champions_league.bronze.partidos_bronze
WHERE
    STATUS = 'FINISHED'
```

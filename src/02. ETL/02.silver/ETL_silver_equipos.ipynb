```sql
INSERT INTO proyecto_champions_league.silver.equipos_silver(id, nombre, sigla, codigo_pais, pais, latitud_pais, longitud_pais, estadio) 
SELECT DISTINCT
   CAST(equipo_id AS INT) as id,
    TRIM( equipo_nombre ) AS nombre,
    TRIM( sigla ) AS sigla,
    TRIM( codigo_pais ) AS codigo_pais ,
    CASE
        WHEN TRIM(pais) = 'Germany' THEN 'Alemania'
        WHEN TRIM(pais) = 'England' THEN 'Inglaterra'
        WHEN TRIM(pais) = 'Spain' THEN 'España'
        WHEN TRIM(pais) = 'Italy' THEN 'Italia'
        WHEN TRIM(pais) = 'Portugal' THEN 'Portugal'
        WHEN TRIM(pais) = 'France' THEN 'Francia'
        WHEN TRIM(pais) = 'Monaco' THEN 'Mónaco'
        WHEN TRIM(pais) = 'Turkey' THEN 'Turquía'
        WHEN TRIM(pais) = 'Azerbaijan' THEN 'Azerbaiyán'
        WHEN TRIM(pais) = 'Greece' THEN 'Grecia'
        WHEN TRIM(pais) = 'Netherlands' THEN 'Países Bajos'
        WHEN TRIM(pais) = 'Scotland' THEN 'Escocia'
        WHEN TRIM(pais) = 'Croatia' THEN 'Croacia'
        WHEN TRIM(pais) = 'Belgium' THEN 'Bélgica'
        WHEN TRIM(pais) = 'Czech Republic' THEN 'República Checa'
        WHEN TRIM(pais) = 'Switzerland' THEN 'Suiza'
        WHEN TRIM(pais) = 'Denmark' THEN 'Dinamarca'
        WHEN TRIM(pais) = 'Austria' THEN 'Austria'
        WHEN TRIM(pais) = 'Ukraine' THEN 'Ucrania'
        WHEN TRIM(pais) = 'Norway' THEN 'Noruega'
        WHEN TRIM(pais) = 'Serbia' THEN 'Serbia'
        WHEN TRIM(pais) = 'Slovakia' THEN 'Eslovaquia'
        WHEN TRIM(pais) = 'Kazakhstan' THEN 'Kazajistán'
        WHEN TRIM(pais) = 'Cyprus' THEN 'Chipre'
        ELSE TRIM(pais)
    END AS pais,
    CASE
        WHEN TRIM(pais) = 'Germany' THEN 51.1657
        WHEN TRIM(pais) = 'England' THEN 52.3555
        WHEN TRIM(pais) = 'Spain' THEN 40.4637
        WHEN TRIM(pais) = 'Italy' THEN 41.8719
        WHEN TRIM(pais) = 'Portugal' THEN 39.3999
        WHEN TRIM(pais) = 'France' THEN 46.6034
        WHEN TRIM(pais) = 'Monaco' THEN 43.7384
        WHEN TRIM(pais) = 'Turkey' THEN 38.9637
        WHEN TRIM(pais) = 'Azerbaijan' THEN 40.1431
        WHEN TRIM(pais) = 'Greece' THEN 39.0742
        WHEN TRIM(pais) = 'Netherlands' THEN 52.1326
        WHEN TRIM(pais) = 'Scotland' THEN 56.4907
        WHEN TRIM(pais) = 'Croatia' THEN 45.1
        WHEN TRIM(pais) = 'Belgium' THEN 50.5039
        WHEN TRIM(pais) = 'Czech Republic' THEN 49.8175
        WHEN TRIM(pais) = 'Switzerland' THEN 46.8182
        WHEN TRIM(pais) = 'Denmark' THEN 56.2639
        WHEN TRIM(pais) = 'Austria' THEN 47.5162
        WHEN TRIM(pais) = 'Ukraine' THEN 48.3794
        WHEN TRIM(pais) = 'Norway' THEN 60.472
        WHEN TRIM(pais) = 'Serbia' THEN 44.0165
        WHEN TRIM(pais) = 'Slovakia' THEN 48.669
        WHEN TRIM(pais) = 'Kazakhstan' THEN 48.0196
        WHEN TRIM(pais) = 'Cyprus' THEN 35.1264
        ELSE NULL
    END AS latitud_pais,
    CASE
        WHEN TRIM(pais) = 'Germany' THEN 10.4515
        WHEN TRIM(pais) = 'England' THEN -1.1743
        WHEN TRIM(pais) = 'Spain' THEN -3.7492
        WHEN TRIM(pais) = 'Italy' THEN 12.5674
        WHEN TRIM(pais) = 'Portugal' THEN -8.2245
        WHEN TRIM(pais) = 'France' THEN 1.8883
        WHEN TRIM(pais) = 'Monaco' THEN 7.4246
        WHEN TRIM(pais) = 'Turkey' THEN 35.2433
        WHEN TRIM(pais) = 'Azerbaijan' THEN 47.5769
        WHEN TRIM(pais) = 'Greece' THEN 21.8243
        WHEN TRIM(pais) = 'Netherlands' THEN 5.2913
        WHEN TRIM(pais) = 'Scotland' THEN -4.2026
        WHEN TRIM(pais) = 'Croatia' THEN 15.2
        WHEN TRIM(pais) = 'Belgium' THEN 4.4699
        WHEN TRIM(pais) = 'Czech Republic' THEN 15.473
        WHEN TRIM(pais) = 'Switzerland' THEN 8.2275
        WHEN TRIM(pais) = 'Denmark' THEN 9.5018
        WHEN TRIM(pais) = 'Austria' THEN 14.5501
        WHEN TRIM(pais) = 'Ukraine' THEN 31.1656
        WHEN TRIM(pais) = 'Norway' THEN 8.4689
        WHEN TRIM(pais) = 'Serbia' THEN 21.0059
        WHEN TRIM(pais) = 'Slovakia' THEN 19.699
        WHEN TRIM(pais) = 'Kazakhstan' THEN 66.9237
        WHEN TRIM(pais) = 'Cyprus' THEN 33.4299
        ELSE NULL
    END AS longitud_pais,
    TRIM( estadio ) AS estadio
FROM proyecto_champions_league.bronze.equipos_bronze
ORDER BY id
```

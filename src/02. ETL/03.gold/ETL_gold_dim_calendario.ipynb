```sql
INSERT INTO proyecto_champions_league.gold.dim_calendario_gold (
  id, fecha,
  anio, anio_mes, anio_mes_num,
  mes, mes_num,
  trimestre, 
  dia_semana, dia_semana_num, es_fin_de_semana
)
SELECT
  CAST( DATE_FORMAT(fecha, 'yyyyMMdd') AS BIGINT ) AS id,
  CAST( fecha AS DATE) AS fecha,
  YEAR(fecha) AS anio,
  DATE_FORMAT(fecha, 'yy-MMM') AS anio_mes,
  CAST( DATE_FORMAT(fecha, 'yyMM') AS INT) AS anio_mes_num,
  DATE_FORMAT(fecha, 'MMM') AS mes,
  MONTH(fecha) AS mes_num,
  QUARTER(fecha) AS trimestre,
  DATE_FORMAT(fecha, 'EEE') AS dia_semana,
  EXTRACT(DAYOFWEEK_ISO FROM fecha) AS dia_semana_num,
  CASE
    WHEN EXTRACT(DAYOFWEEK_ISO FROM fecha) IN (6,7) THEN TRUE
    ELSE FALSE
  END AS es_fin_de_semana
FROM (
    SELECT sequence(
        DATE_ADD(DATE_TRUNC('YEAR', CURRENT_DATE()), -5 * 365),
        DATE_ADD(DATE_TRUNC('YEAR', CURRENT_DATE()), 5 * 365),
        INTERVAL 1 DAY
    ) as fechas
) tmp
LATERAL VIEW explode(fechas) as fecha
ORDER BY fecha;
```

```sql
INSERT INTO proyecto_champions_league.gold.fact_partidos_gold(
    id,
    temporada,
    fecha,
    fase,
    grupo,
    jornada,
    local_id,
    visitante_id,
    arbitro_id,
    ganador,
    goles_local,
    goles_visitante,
    goles_local_ht,
    goles_visitante_ht,
    goles_local_et,
    goles_visitante_et,
    goles_local_pen,
    goles_visitante_pen
)
SELECT
    part.id,
    part.temporada,
    part.fecha,
    part.fase,
    part.grupo,
    part.jornada,
    part.local_id,
    part.visitante_id,
    arb.id AS arbitro_id,
    part.ganador,
    part.goles_local,
    part.goles_visitante,
    part.goles_local_ht,
    part.goles_visitante_ht,
    part.goles_local_et,
    part.goles_visitante_et,
    part.goles_local_pen,
    part.goles_visitante_pen
FROM proyecto_champions_league.silver.partidos_silver part
LEFT JOIN proyecto_champions_league.gold.dim_arbitros_gold arb ON part.arbitro = arb.arbitro
ORDER BY part.fecha DESC
```

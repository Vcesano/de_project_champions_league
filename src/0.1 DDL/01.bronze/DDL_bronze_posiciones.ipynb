```sql
DROP TABLE IF EXISTS proyecto_champions_league.bronze.posiciones_bronze;
CREATE TABLE IF NOT EXISTS proyecto_champions_league.bronze.posiciones_bronze (
    temporada STRING,
    grupo STRING,
    posicion STRING,
    equipo_id STRING,
    equipo_nombre STRING,
    jugados STRING,
    ganados STRING,
    empatados STRING,
    perdidos STRING,
    goles_favor STRING,
    goles_contra STRING,
    diferencia STRING,
    puntos STRING,
    _processing_timestamp TIMESTAMP DEFAULT current_timestamp() COMMENT 'Timestamp de procesamiento'
)
USING DELTA
TBLPROPERTIES ('delta.feature.allowColumnDefaults' = 'supported')
COMMENT 'Tabla de posiciones - BRONZE'
```
